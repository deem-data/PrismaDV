```python
import pandas as pd

from workflow_prismadv.results import methods
from workflow_prismadv.results.metrics import dataset_summary
from workflow_prismadv.results.per_dataset_table import DATASETS, LABELS
```

```python
outcomes = methods.outcomes(benchmarks=list(DATASETS))
```

```python
def table(metric="f1"):
    ds = dataset_summary(outcomes)

    def cell(row):
        v = f"{100 * row[f'{metric}_mean']:.1f}"
        if row.reps >= 2:
            v += f" ± {100 * row[f'{metric}_sd']:.1f}"
        return v

    return pd.DataFrame({(b, LABELS.get(d, d)): {r.name: cell(ds.loc[(b, r.key, d)]) for r in methods.ROWS}
                         for b, dss in DATASETS.items() for d in dss})
```

```python
# Table J: macro F1 per dataset, mean ± SD over three runs
table("f1")
```
