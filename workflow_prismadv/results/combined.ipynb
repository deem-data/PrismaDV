```python
import pandas as pd

from workflow_prismadv.results import methods
from workflow_prismadv.results.metrics import METRICS, ci_half_width, summary
```

```python
outcomes = methods.outcomes()
```

```python
def table(kind="macro", interval="sd"):
    s = summary(outcomes, kind)

    def cell(row, m):
        v = f"{100 * row[f'{m}_mean']:.1f}"
        if row.reps >= 2:
            sd = row[f"{m}_sd"]
            w = sd if interval == "sd" else ci_half_width(sd, int(row.reps))
            v += f" ± {100 * w:.1f}"
        return v

    return pd.DataFrame({(b, m): {r.name: cell(s.loc[(b, r.key)], m) for r in methods.ROWS}
                         for b in methods.BENCHMARKS for m in METRICS})
```

```python
# Table 1: macro precision, recall and F1, mean ± SD over three runs
table("macro", "sd")
```

```python
# Table E: macro scores, mean ± half-width of the 95% t confidence interval
table("macro", "ci95")
```

```python
# Table H: binary scores, safe = positive class, mean ± SD
table("binary", "sd")
```
