```python
import pandas as pd

from workflow_prismadv.results import ablation
from workflow_prismadv.results.benchmarks import real
from workflow_prismadv.results.metrics import METRICS, dataset_summary, summary
from workflow_prismadv.results.per_dataset_table import LABELS
```

```python
outcomes = ablation.outcomes()
ds, s = dataset_summary(outcomes), summary(outcomes)
```

```python
def cell(row, m):
    return f"{100 * row[f'{m}_mean']:.1f} ± {100 * row[f'{m}_sd']:.1f}"


rows = {}
for key, name in ablation.ROWS.items():
    for d in ["bank_marketing_analysis", "healthy_diet_dashboard", "omop_cdm_databricks"]:
        rows[(name, LABELS[d])] = {m: cell(ds.loc[("real", key, d)], m) for m in METRICS}
    rows[(name, "Mean")] = {m: cell(s.loc[("real", key)], m) for m in METRICS}

# Table 2: PrismaDV ablation on EIDBench-real, macro scores, mean ± SD over three runs
pd.DataFrame(rows).T
```
