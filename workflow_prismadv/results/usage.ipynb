```python
import pandas as pd

from workflow_prismadv.results import methods, usage
from workflow_prismadv.results.usage_table import COLUMNS
```

```python
s = usage.summary(usage.records(["synth"]))
```

```python
def cell(row, field):
    return f"{row[f'{field}_mean'] / 1000:.1f} ± {row[f'{field}_sd'] / 1000:.1f}"


# Table F: mean tokens per task on EIDBench-synth (thousands), ± SD over three runs
pd.DataFrame({h: {r.name: cell(s.loc[("synth", r.key)], f) for r in methods.ROWS if r.key in usage.ARMS}
              for h, f in COLUMNS})
```
